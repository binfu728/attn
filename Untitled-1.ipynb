```python
with h5py.File("/mnt/ht2-nas2/00-model/00-fb/olmo_test/inference_data/sample_0.h5", 'r') as f:
    dataset = f['landsat'][:]
    ts = f["missing_timesteps_masks"]
    for name in ts.keys():
        data = ts[name][:]   # 读取整个数据集，得到 NumPy 数组
        print(f"{name}: {data}")  # 直接打印数组内容
    for time_idx in range(12):
        for band_idx in range(11):
            img = dataset[:, :, time_idx, band_idx]
            clear_output(wait=True)        # 清除之前的输出
            plt.imshow(img, cmap='gray')
            plt.title(f'Band {band_idx}, Time {time_idx}')
            plt.show()
            # 可以加一个短暂延时，避免过快
            time.sleep(0.2)
```
